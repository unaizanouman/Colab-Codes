```python
from math import *

h=0.001
x=-pi

while x<=pi:

    approx=(sin(x+h)-sin(x))/h
    actual=cos(x)

    print("x =", round(x,3),
          " Approx =",round(approx,5),
          " cos(x) =",round(actual,5))

    x=x+0.001
```

```python
birthdays={
    "Albert Einstein": "14/03/1879",
    "Benjamin Franklin": "17/01/1706",
    "Ada Lovelace": "10/12/1815"
}

print("Welcome to the birthday dictionary.")
print("We know the birthdays of:")

for name in birthdays:
    print(name)
person=input("Whose birthday do you want to look up? ")

if person in birthdays:
    print(person+"'s birthday is",birthdays[person])
else:
    print("Sorry, birthday not found.")
```
